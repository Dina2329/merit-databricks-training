## Widgets

```python
dbutils.widgets.help()
```

Combodox we can use any value not in choice also.

Dropdown means we can only choose from choice.


Text any thing can enter.

```python
dbutils.widgets.combobox(name="top_n",defaultValue="1", choices=["1", "2", "3"])
dbutils.widgets.dropdown(name="Network",defaultValue="5G",choices=["5G", "4G","3G","2G"],label="Technology Type")
dbutils.widgets.text(name="text",defaultValue="text", label="Customer_name")
```

```python
dbutils.widgets.get(name="top_n")
```

```python
dbutils.widgets.get(name="Network")
```

```python
dbutils.widgets.getAll()
```
